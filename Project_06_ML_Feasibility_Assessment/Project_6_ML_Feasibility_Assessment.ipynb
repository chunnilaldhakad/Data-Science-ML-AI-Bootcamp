# Project 6 - Machine Learning Feasibility Assessment

## Objective

Evaluate whether Machine Learning is an appropriate solution for the customer churn problem before developing predictive models.

### Learning Outcomes
- Understand the business problem
- Assess dataset suitability
- Evaluate ML feasibility
- Select candidate algorithms
- Identify risks and success criteria


## Business Problem

A telecom company wants to identify customers who are likely to churn so it can proactively retain them.

This project determines whether Machine Learning is the right solution for this problem.


```
import pandas as pd

df = pd.read_csv("cleaned_data/telco_cleaned.csv")

print("Dataset Shape:", df.shape)
df.head()
```

## Dataset Quality Assessment

```
report = {
    "Rows": df.shape[0],
    "Columns": df.shape[1],
    "Missing Values": int(df.isnull().sum().sum()),
    "Duplicate Rows": int(df.duplicated().sum()),
    "Target Available": "Churn" in df.columns
}

for k,v in report.items():
    print(f"{k}: {v}")
```

## Is Machine Learning Suitable?

```
criteria = {
    "Historical Data Available": True,
    "Target Variable Available": "Churn" in df.columns,
    "Enough Samples (>1000)": df.shape[0] > 1000,
    "Structured Dataset": True,
    "Prediction Required": True
}

feasible = all(criteria.values())

for k,v in criteria.items():
    print(f"{k}: {v}")

print("\nML Feasible:", feasible)
```

## Problem Type

```
target = "Churn"

if target in df.columns:
    print("Problem Type : Binary Classification")
    print(df[target].value_counts())
else:
    print("Target variable not found.")
```

## Candidate Algorithms

```
algorithms = [
    "Logistic Regression",
    "Decision Tree",
    "Random Forest",
    "Support Vector Machine",
    "K-Nearest Neighbors",
    "XGBoost",
    "LightGBM",
    "CatBoost"
]

for algo in algorithms:
    print("-", algo)
```

## Algorithm Selection Matrix

```
selection = pd.DataFrame({
    "Algorithm":[
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "XGBoost"
    ],
    "Interpretability":["High","Medium","Medium","Low"],
    "Training Speed":["Fast","Fast","Medium","Medium"],
    "Expected Performance":["Good","Good","Very Good","Excellent"]
})

selection
```

## Risks & Limitations

```
risks = [
    "Class imbalance",
    "Data leakage",
    "Overfitting",
    "Changing customer behavior",
    "Missing values",
    "Feature drift"
]

for r in risks:
    print("-", r)
```

## Recommendation

Machine Learning **is recommended** because:

- Historical labeled data is available.
- The target variable (Churn) exists.
- The dataset is structured.
- The problem is a supervised binary classification task.
- Business value from prediction is high.


# Conclusion

The feasibility assessment confirms that Machine Learning is an appropriate approach.
