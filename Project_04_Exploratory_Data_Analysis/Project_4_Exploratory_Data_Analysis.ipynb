# Project 4 - Exploratory Data Analysis (EDA)

This notebook performs Exploratory Data Analysis (EDA) on the cleaned Telco Customer Churn dataset to uncover trends, patterns, relationships, and business insights.

## Objectives
- Load cleaned dataset
- Understand distributions
- Perform univariate, bivariate, and multivariate analysis
- Visualize data
- Generate business insights

```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cleaned_data/telco_cleaned.csv")
df.head()
```

```
print("Shape:", df.shape)
df.info()
```

## Summary Statistics

```
df.describe(include='all').T
```

## Missing Values

```
df.isnull().sum()
```

## Univariate Analysis

```
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=30)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

## Boxplots for Numerical Features

```
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.boxplot(df[col].dropna(), vert=True)
    plt.title(f"Boxplot - {col}")
    plt.show()
```

## Categorical Analysis

```
categorical_cols = df.select_dtypes(exclude='number').columns

for col in categorical_cols:
    plt.figure(figsize=(8,4))
    df[col].value_counts().plot(kind='bar')
    plt.title(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

## Correlation Matrix

```
corr = df.select_dtypes(include='number').corr()

plt.figure(figsize=(8,6))
plt.imshow(corr)
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.colorbar()
plt.title("Correlation Matrix")
plt.show()
```

## Churn Analysis

```
if "Churn" in df.columns:
    print(df["Churn"].value_counts())

    df["Churn"].value_counts().plot(kind="bar")
    plt.title("Customer Churn Distribution")
    plt.ylabel("Customers")
    plt.show()
```
