# Project 2 - Python-Based Data Processing

Load and inspect the dataset using Python and Pandas.

## Objectives
- Load dataset
- Inspect structure
- Process basic data
- Save processed dataset

```
from pathlib import Path
import pandas as pd

data_path = Path("data/Telco-Customer-Churn.csv")
if not data_path.is_file():
    raise FileNotFoundError(f"Dataset not found: {data_path}. Place the CSV in the data/ directory.")

df = pd.read_csv(data_path)
df.head()
```

```
df.info()
```

```
print(f"Rows: {len(df):,}")
print(f"Columns: {df.shape[1]}")
assert not df.empty, "The loaded dataset is empty."
assert df.columns.is_unique, "Dataset contains duplicate column names."
print("Basic dataset checks passed.")
```

```
print(df.shape)
print(df.columns.tolist())
```

```
df.describe(include="all").T
```

```
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

```
obj_cols=df.select_dtypes(include='object').columns
for c in obj_cols:
    df[c]=df[c].astype(str).str.strip()
if 'TotalCharges' in df.columns:
    df['TotalCharges']=pd.to_numeric(df['TotalCharges'], errors='coerce')
df.head()
```

```
import os
os.makedirs("processed_data", exist_ok=True)
df.to_csv("processed_data/telco_processed.csv", index=False)
print("Processed dataset saved.")
```
