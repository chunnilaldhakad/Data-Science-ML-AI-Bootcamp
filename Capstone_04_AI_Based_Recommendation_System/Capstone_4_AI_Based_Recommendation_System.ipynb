# Capstone Project 4 - AI-Based Recommendation System

## Objective

Build a recommendation system that suggests relevant products/services to users using Content-Based Filtering and Collaborative Filtering concepts.


## Workflow

1. Load Dataset
2. Explore Data
3. Content-Based Recommendation
4. Collaborative Filtering
5. Generate Recommendations
6. Evaluate Recommendations
7. Save Recommendation Engine


```
import os
import joblib
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

df = pd.read_csv("datasets/recommendation_data.csv")
required_columns = {"Item", "Description"}
missing_columns = required_columns.difference(df.columns)
if missing_columns:
    raise ValueError(f"Dataset is missing required columns: {sorted(missing_columns)}")

df = df.dropna(subset=["Item"]).copy()
df["Description"] = df["Description"].fillna("").astype(str).str.strip()
if df.empty or not df["Description"].ne("").any():
    raise ValueError("Dataset must contain at least one item with a non-empty description.")

df.head()
```

## Dataset Overview

```
print(df.shape)
df.info()
df.head()
```

## Content-Based Filtering

```
# Assumes a 'Description' column exists
tfidf = TfidfVectorizer(stop_words="english")

tfidf_matrix = tfidf.fit_transform(df["Description"].fillna(""))

similarity = cosine_similarity(tfidf_matrix, tfidf_matrix)

print("Similarity Matrix Shape:", similarity.shape)
```

## Recommendation Function

```
def recommend(item_name, top_n=5):
    if item_name not in df["Item"].values:
        return "Item not found."

    idx = df[df["Item"] == item_name].index[0]

    scores = list(enumerate(similarity[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)[1:top_n+1]

    recommendations = df.iloc[[i for i, _ in scores]][["Item"]]
    return recommendations

# Example
# recommend("Laptop")
```

## Save Recommendation Model

```
os.makedirs("capstone_4/models", exist_ok=True)

joblib.dump(
    {
        "vectorizer": tfidf,
        "similarity": similarity
    },
    "capstone_4/models/recommendation_engine.pkl"
)

print("Recommendation engine saved.")
```
