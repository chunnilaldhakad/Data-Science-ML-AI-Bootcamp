# Capstone Project 2 - Customer Segmentation & Profiling

## Objective

Build an end-to-end customer segmentation system using unsupervised machine learning to identify meaningful customer groups and provide actionable business strategies.


## Workflow

1. Load Dataset
2. Data Preparation
3. Feature Encoding
4. Feature Scaling
5. Elbow Method
6. K-Means Clustering
7. Silhouette Score
8. PCA Visualization
9. Customer Profiling
10. Business Recommendations


```
import os
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

df = pd.read_csv("cleaned_data/telco_cleaned.csv")
df.head()
```

## Data Preparation

```
data = df.copy()

for col in data.select_dtypes(include="object").columns:
    if col != "Churn":
        data[col] = LabelEncoder().fit_transform(data[col].astype(str))

if "Churn" in data.columns:
    data = data.drop(columns=["Churn"])

scaler = StandardScaler()
X = scaler.fit_transform(data)

print("Prepared Shape:", X.shape)
```

## Find Optimal Clusters (Elbow Method)

```
inertia = []

for k in range(2,11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X)
    inertia.append(model.inertia_)

plt.figure(figsize=(6,4))
plt.plot(range(2,11), inertia, marker="o")
plt.xlabel("Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid(True)
plt.show()
```

## Train Final K-Means Model

```
k = 4

kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X)

data["Cluster"] = clusters

print("Silhouette Score:", silhouette_score(X, clusters))
data["Cluster"].value_counts()
```

## PCA Visualization

```
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.figure(figsize=(7,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=clusters)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Customer Segments")
plt.show()
```

## Customer Profiling

```
profiles = data.groupby("Cluster").mean(numeric_only=True)
profiles
```

## Save Outputs

```
os.makedirs("capstone_2/output", exist_ok=True)

data.to_csv("capstone_2/output/customer_segments.csv", index=False)
profiles.to_csv("capstone_2/output/customer_profiles.csv")

print("Segmentation results saved.")
```
