```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

df = pd.read_csv("cleaned_data/telco_cleaned.csv")

df.head()
```

## Dataset Overview

```
print("Shape:", df.shape)
df.info()
```

## Descriptive Statistics

```
df.describe().T
```

## Mean, Median, Mode

```
numeric = df.select_dtypes(include=np.number).columns

stats_df = pd.DataFrame({
    "Mean": df[numeric].mean(),
    "Median": df[numeric].median(),
    "Mode": df[numeric].mode().iloc[0]
})

stats_df
```

## Variance and Standard Deviation

```
variance = df[numeric].var()
std = df[numeric].std()

pd.DataFrame({
    "Variance": variance,
    "Std Dev": std
})
```

## Correlation Analysis

```
corr = df[numeric].corr()

plt.figure(figsize=(8,6))
plt.imshow(corr)
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.colorbar()
plt.title("Correlation Matrix")
plt.show()
```

## Confidence Interval Example

```
monthly = df["MonthlyCharges"].dropna()

confidence = 0.95
mean = monthly.mean()
sem = stats.sem(monthly)

interval = stats.t.interval(confidence,
                            len(monthly)-1,
                            loc=mean,
                            scale=sem)

print("95% Confidence Interval")
print(interval)
```

## Hypothesis Testing

```
if "Churn" in df.columns:

    churn_yes = df[df["Churn"]=="Yes"]["MonthlyCharges"]
    churn_no = df[df["Churn"]=="No"]["MonthlyCharges"]

    t_stat, p_value = stats.ttest_ind(
        churn_yes,
        churn_no,
        equal_var=False,
        nan_policy="omit"
    )

    print("T Statistic:", t_stat)
    print("P-value:", p_value)

    alpha = 0.05

    if p_value < alpha:
        print("Reject Null Hypothesis")
    else:
        print("Fail to Reject Null Hypothesis")
```
