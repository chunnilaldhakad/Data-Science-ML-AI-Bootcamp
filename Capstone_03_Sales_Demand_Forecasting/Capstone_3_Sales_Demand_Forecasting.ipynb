# Capstone Project 3 - Sales / Demand Forecasting

## Objective

Develop a time-series forecasting solution to predict future sales or demand using historical data and evaluate forecasting performance.


## Workflow

1. Load Dataset
2. Convert Date Column
3. Aggregate Sales by Date
4. Visualize Trend
5. Train-Test Split
6. Build Forecasting Model
7. Evaluate Forecast
8. Predict Future Demand
9. Save Model & Forecast


```
import os
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.holtwinters import ExponentialSmoothing
import numpy as np

# Replace with your sales dataset
df = pd.read_csv("datasets/sales_data.csv")
df.head()
```

## Prepare Time Series

```
df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date")

sales = df.groupby("Date")["Sales"].sum()

sales.head()
```

## Visualize Sales Trend

```
plt.figure(figsize=(10,5))
plt.plot(sales)
plt.title("Historical Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

## Train-Test Split

```
train_size = int(len(sales) * 0.8)

train = sales.iloc[:train_size]
test = sales.iloc[train_size:]

print("Train:", len(train))
print("Test :", len(test))
```

## Train Forecasting Model

```
model = ExponentialSmoothing(
    train,
    trend="add",
    seasonal=None
)

fit = model.fit()

forecast = fit.forecast(len(test))
```

## Model Evaluation

```
mae = mean_absolute_error(test, forecast)
rmse = np.sqrt(mean_squared_error(test, forecast))

print("MAE :", mae)
print("RMSE:", rmse)
```

## Forecast vs Actual

```
plt.figure(figsize=(10,5))
plt.plot(train.index, train, label="Train")
plt.plot(test.index, test, label="Actual")
plt.plot(test.index, forecast, label="Forecast")
plt.legend()
plt.title("Sales Forecast")
plt.show()
```

## Forecast Next 30 Periods

```
future = fit.forecast(30)

future.head()
```

## Save Forecast

```
os.makedirs("capstone_3/output", exist_ok=True)

future.to_csv("capstone_3/output/future_forecast.csv")

print("Forecast saved.")
```
